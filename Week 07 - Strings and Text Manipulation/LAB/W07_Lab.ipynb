## Topic 1 · Case and whitespace
*Walkthrough: `01 - Case and Whitespace`*

### D1. Case changes comparisons

1. Create `a = "SRV-01"` and `b = "srv-01"`.
2. Print `a == b`.
3. Print the same comparison again, with `.lower()` on both sides.

**Expected output:** `False`, then `True`

````python
# D1
a = "SRV-01"
b = "srv-01"
print(a == b) # compares contents of a with contents of b, outputs a boolean value

print(a.lower() == b.lower()) # returns string in a and b to lower case and compares them, outputs a boolean value
````

### D2. `.upper()`, `.lower()`, `.title()`

1. Create `name = "record check console"` and print it with `.title()`, then with `.upper()`.
2. Create `original = "SRV-01"`. Print `original.lower()`, then print `original` again.
3. Add a comment explaining why `original` did not change.

**Expected output:**
```
Record Check Console
RECORD CHECK CONSOLE
srv-01
SRV-01
```

````python
# D2
name = "record check console"
print(name.title()) # outputs the first letter of each word to upper case
print(name.upper()) # outputs the whole string into upper case
original = "SRV-01"
print(original.lower()) # outputs the whole string into lower case
print(original)
# contents of original do not change as the method lower() only returns the value in lower case, it does not assign the new value directly to the variable
````

### D3. `.strip()`

1. Create `raw = "  srv-01  "`.
2. Print it inside square brackets so you can see the spaces: `print(f"[{raw}]")`
3. Print it again the same way, with `.strip()`.

**Expected output:** `[  srv-01  ]`, then `[srv-01]`

````python
# D3
raw = " srv-01 "
print(f"[{raw}]")
print(f"[{raw.strip()}]") # removes spaces on each ends of the string
````

### D4. `.lstrip()`, `.rstrip()` and chaining

1. Create `messy = "***srv-01***"` and print `messy.strip("*")`, `messy.lstrip("*")` and `messy.rstrip("*")`.
2. Create `field = "  SRV-01  "` and print `field.strip().lower()` (two methods chained together).

**Expected output:** `srv-01`, `srv-01***`, `***srv-01`, then `srv-01`

````python
# D4
messy = "***srv-01***"
print(messy.strip("*"))
print(messy.lstrip("*")) # removes all * from the left end of the string
print(messy.rstrip("*")) # removes all * from the right end of the string
field = " SRV-01 "
print(field.strip().lower()) # removes all spaces from both ends of the string and puts the string in lower case
````

### D8. `.startswith()`, `.endswith()`, `.count()`

1. Create `lines = ["ERROR: disk full", "INFO: backup complete", "ERROR: timeout"]`. Loop over it
   and print `Needs attention: ` plus the line, but only for lines that start with `"ERROR"`.

## Topic 4 · Parsing real text
*Walkthrough: `04 - Parsing Real Text`*

### D14. Parse one messy line, step by step

Start with `raw_line = "  Srv-01 , used=87 , total=100  "`.

1. `.strip()` the whole line.
2. Split off the hostname with `.split(",", 1)` into `hostname_part` and `rest`.
3. Clean the hostname with `.strip().lower()`.
4. Split `rest` on `","`. For each piece, split on `"="`, strip the key and the value, and convert the
   value with `float()`.
5. Put everything into one dict and print it.

**Expected output:** `{'hostname': 'srv-01', 'used': 87.0, 'total': 100.0}`

````python
# D14
"""this program takes a string of record, cleans it and outputs it"""
raw_line = " srv-01 , used=87 , total=100 "
raw_line = raw_line.strip()
hostname_part,rest = raw_line.split(",", 1)
hostname_part = hostname_part.strip().lower()
rest = rest.split(",")
Result = {"hostname" : hostname_part} # initialisation of the dictionary
for i in range(len(rest)): # looping over the "rest" list
    key,value = rest[i].split("=")
    key = key.strip()
    value = value.strip()
    Result[key] = float(value) # adds key value pairs to the dictionary, converts value to float datatype
print(Result)
````

### D15. Turn it into a function

1. Put your steps from D14 into a function `parse_line(raw_line)` that **returns** the dict.
   Give it a one-line docstring.
2. Print `parse_line("SRV-02,used=120,total=100")`.

**Expected output:** `{'hostname': 'srv-02', 'used': 120.0, 'total': 100.0}`

````python
# D15
def parse_line(raw_line):
    """this function takes a string of record, cleans it and returns it"""
    raw_line = raw_line.strip()
    hostname_part,rest = raw_line.split(",", 1)
    hostname_part = hostname_part.strip().lower()
    rest = rest.split(",")
    Result = {"hostname" : hostname_part} # initialisation of the dictionary
    for i in range(len(rest)): # looping over the "rest" list
        key,value = rest[i].split("=")
        key = key.strip()
        value = value.strip()
        Result[key] = float(value) # adds key value pairs to the dictionary, converts value to float datatype
    return Result
print(parse_line("SRV-02,used=120,total=100"))
````

### C1. Parse a whole file

Read `messy_lines.txt` (in this `LAB` folder) line by line. Parse each line with `parse_line` from D15,
and print the hostname with `OVER LIMIT` (100% or more), `WARNING` (90% or more) or `OK`.
Do not write the cleaning logic a second time.

**Expected output:**
```
srv-01 OK
srv-02 OVER LIMIT
srv-03 OK
srv-04 WARNING
```

````python
# C1
"""This program opens a file, reads it line by line and outputs the record label along with its status"""
with open("messy_lines.txt", "r") as f: # opens file in read mode
    for line in f:
        record = parse_line(line) # calls function parse_line with parameter of line read from file
        percent = (record["used"]/record["total"]) * 100 # calculates the percentage to be used in comparison
        if percent >= 100:
           print(f"{record["hostname"]} OVER LIMIT")
        elif percent >= 90:
            print(f"{record["hostname"]} WARNING")
        else:
            print(f"{record["hostname"]} OK")
        
````

Loop over the three lines below. Use `try` / `except ValueError` so that lines that do not parse print
`Could not parse:` and the line, instead of crashing.

```python
lines = ["srv-07,used=50,total=100",
         "srv-08,used=unknown,total=100",
```
